# Criar e Acessar Banco de Dados SQLite usando Python

Tempo estimado necessário: **15** minutos

## Objetivos

Após completar este laboratório, você será capaz de:

* Criar um banco de dados
* Criar uma tabela
* Inserir dados na tabela
* Consultar dados da tabela
* Recuperar o conjunto de resultados em um dataframe pandas
* Fechar a conexão com o banco de dados


SQLite é uma biblioteca de software que implementa um mecanismo de banco de dados SQL transacional, autônomo, sem servidor e de configuração zero. SQLite é o mecanismo de banco de dados SQL mais amplamente implantado no mundo.

## Tarefa 1: Criar banco de dados usando SQLite


````python
# Instalar e carregar sqlite3

#!pip install sqlite3  ##Descomente este código apenas se você estiver trabalhando em um ambiente local para instalar o sqlite3
import sqlite3
````

````python
# Conectando ao sqlite
# objeto de conexão
conn = sqlite3.connect('INSTRUCTOR.db')
````

A classe Cursor é uma instância usando a qual você pode invocar métodos que executam instruções SQLite e buscam dados dos conjuntos de resultados das consultas. Você pode criar o objeto Cursor usando o método cursor() do objeto/classe Connection.


````python
# objeto cursor
cursor_obj = conn.cursor()
````

## Tarefa 2: Criar uma tabela no banco de dados

Nesta etapa, criaremos uma tabela no banco de dados com os seguintes detalhes:

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/images/table.png" align="center">


Antes de criar uma tabela, vamos primeiro verificar se a tabela já existe ou não. Para descartar a tabela de um banco de dados, use a consulta **DROP**. Um cursor é um objeto que ajuda a executar a consulta e buscar os registros do banco de dados.


````python
# Remove a tabela se ela já existir.
cursor_obj.execute("DROP TABLE IF EXISTS INSTRUCTOR")
````

## Não se preocupe se você receber este erro:

Se você vir uma exceção/erro semelhante ao seguinte, indicando que INSTRUCTOR é um nome indefinido, tudo bem. Isso apenas implica que a tabela INSTRUCTOR não existe na tabela - o que seria o caso se você não a tivesse criado anteriormente.

<strong>Exception: \[IBM]\[CLI Driver]\[DB2/LINUXX8664] SQL0204N  "ABC12345.INSTRUCTOR" is an undefined name.  SQLSTATE=42704 SQLCODE=-204</strong>


````python
# Criando a tabela
table = """ create table IF NOT EXISTS INSTRUCTOR(ID INTEGER PRIMARY KEY NOT NULL, FNAME VARCHAR(20), LNAME VARCHAR(20), CITY VARCHAR(20), CCODE CHAR(2));"""
 
cursor_obj.execute(table)
 
print("Table is Ready")
````

## Tarefa 3: Inserir dados na tabela

Nesta etapa, inseriremos algumas linhas de dados na tabela.

A tabela INSTRUCTOR que criamos na etapa anterior contém 3 linhas de dados:

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/images/table1.png" align="center">

Começaremos inserindo apenas a primeira linha de dados, ou seja, para o instrutor Rav Ahuja


````python
cursor_obj.execute('''insert into INSTRUCTOR values (1, 'Rav', 'Ahuja', 'TORONTO', 'CA')''')
````

A saída que você obterá será algo como: <strong>sqlite3.Cursor at 0x27a1a491260</strong>, o que significa que o banco de dados mySql tem o objeto sqlite3.Cursor em 0x27a1a49126 como saída na tabela. Mas você pode obter um número diferente.

Agora use uma única consulta para inserir as duas linhas de dados restantes


````python
cursor_obj.execute('''insert into INSTRUCTOR values (2, 'Raul', 'Chong', 'Markham', 'CA'), (3, 'Hima', 'Vasudevan', 'Chicago', 'US')''')
````

## Tarefa 4: Consultar dados na tabela

Nesta etapa, recuperaremos os dados que inserimos na tabela INSTRUCTOR.


````python
statement = '''SELECT * FROM INSTRUCTOR'''
cursor_obj.execute(statement)

print("All the data")
output_all = cursor_obj.fetchall()
for row_all in output_all:
  print(row_all)
````

````python
## Buscar algumas linhas da tabela
statement = '''SELECT * FROM INSTRUCTOR'''
cursor_obj.execute(statement)
  
print("All the data")
# Se você quiser buscar algumas linhas da tabela, usamos fetchmany(numerodelinhas) e mencionamos o número de linhas que você deseja buscar
output_many = cursor_obj.fetchmany(2) 
for row_many in output_many:
  print(row_many)
````

````python
# Buscar apenas FNAME da tabela
statement = '''SELECT FNAME FROM INSTRUCTOR'''
cursor_obj.execute(statement)
  
print("All the data")
output_column = cursor_obj.fetchall()
for fetch in output_column:
  print(fetch)
````

````python
query_update='''update INSTRUCTOR set CITY='MOOSETOWN' where FNAME="Rav"'''
cursor_obj.execute(query_update)
````
